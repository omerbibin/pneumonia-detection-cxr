```python
# A clean folder structure now will save us hours of confusion later.
# We follow a layout commonly used in MLOps / Kaggle projects.

import os  # Standard library for file system operations

# Root project folder lives in your Google Drive so it survives between Colab sessions.
PROJECT_ROOT = "/content/drive/MyDrive/PneumoniaDetection"

# Sub-folders for each concern. Keeping data, code, models, and reports separated
# is a CRISP-DM best practice and makes the GitHub repo easy to navigate.
SUBFOLDERS = [
    "data/raw",          # Original, unmodified dataset (read-only after download)
    "data/processed",    # Cleaned / resized images ready for training
    "notebooks",         # One notebook per week (Week01_..., Week02_..., etc.)
    "src",               # Reusable Python modules we will write later
    "models",            # Saved model weights (.h5 / .keras files)
    "reports/figures",   # Plots, confusion matrices, ROC curves
    "reports/literature" # PDFs of papers we read in the literature review
]

# Create each folder if it doesn't already exist. exist_ok=True prevents errors on reruns.
for sub in SUBFOLDERS:
    full_path = os.path.join(PROJECT_ROOT, sub)
    os.makedirs(full_path, exist_ok=True)
    print("Created or verified:", full_path)
```

```python
# We will keep a simple text log of weekly progress. This is invaluable when writing
# the final IEEE report in Week 12, because you will not remember details otherwise.

log_path = os.path.join(PROJECT_ROOT, "PROGRESS_LOG.md")  # Markdown log file

# Only write the initial header if the log doesn't exist yet, so we don't overwrite later entries.
if not os.path.exists(log_path):
    with open(log_path, "w", encoding="utf-8") as f:
        f.write("# Pneumonia Detection — Progress Log\n\n")
        f.write("## Week 1 — Project Setup\n")
        f.write("- Verified Colab environment (Python, TF, GPU).\n")
        f.write("- Mounted Google Drive.\n")
        f.write("- Created project folder structure.\n")
        f.write("- Started literature review (see reports/literature/).\n\n")
    print("Progress log initialized at:", log_path)
else:
    print("Progress log already exists at:", log_path)
```
